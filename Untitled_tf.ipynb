```python
from sklearn.datasets import fetch_20newsgroups
import pandas as pd
```

```python
newsgroups = fetch_20newsgroups(
    subset="all",
    remove=("headers", "footers", "quotes"),  # نشيل التوقيعات والاقتباسات عشان مايأثروش على التجميع
    random_state=42
)
```

```python
df = pd.DataFrame({
    "text": newsgroups.data,
    "category": [newsgroups.target_names[t] for t in newsgroups.target]
})

print(df.shape)
print(df["category"].value_counts())
print(df["text"].iloc[0][:500])  # أول 500 حرف من أول مستند
```

```python
import re

def clean_text(text):
    text = text.lower()                          # توحيد الحالة (كبيرة/صغيرة)
    text = re.sub(r"\S+@\S+", " ", text)          # شيل الإيميلات
    text = re.sub(r"http\S+|www\S+", " ", text)   # شيل الروابط
    text = re.sub(r"[^a-z\s]", " ", text)         # شيل أي حاجة مش حروف إنجليزية
    text = re.sub(r"\s+", " ", text).strip()      # شيل المسافات الزيادة
    return text

df["clean_text"] = df["text"].apply(clean_text)
```

```python
# نشيل أي مستند بقى فاضي أو قصير جدًا بعد التنضيف (مفيدش معلومة)
df["text_length"] = df["clean_text"].str.split().apply(len)
print(df["text_length"].describe())
```

```python
df = df[df["text_length"] >= 20]  # نحتفظ بالمستندات اللي فيها 20 كلمة على الأقل
print(df.shape)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")

# 1) توزيع الفئات (عشان نتأكد بصريًا من التوازن)
plt.figure(figsize=(10, 6))
df["category"].value_counts().plot(kind="barh")
plt.title("Document Count per Category")
plt.xlabel("Count")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()

# 2) توزيع طول المستندات (قبل التنضيف المتطرف)
fig, axes = plt.subplots(1, 2, figsize=(14, 4))
axes[0].hist(df["text_length"], bins=50, range=(0, 500))
axes[0].set_title("Document Length (words) — zoomed 0-500")
axes[1].boxplot(df["text_length"], vert=False)
axes[1].set_title("Document Length — Boxplot (shows outliers)")
plt.tight_layout()
plt.show()

# 3) متوسط طول المستند حسب الفئة
avg_len = df.groupby("category")["text_length"].median().sort_values()
plt.figure(figsize=(10, 6))
avg_len.plot(kind="barh")
plt.title("Median Document Length by Category")
plt.tight_layout()
plt.show()

# 4) أكتر 20 كلمة تكرارًا في كل الداتا (بعد التنضيف، قبل TF-IDF)
from collections import Counter
all_words = " ".join(df["clean_text"]).split()
common_words = Counter(all_words).most_common(30)
words, counts = zip(*common_words)

plt.figure(figsize=(10, 8))
plt.barh(words[::-1], counts[::-1])
plt.title("Top 30 Most Frequent Words (raw, before stopword removal)")
plt.tight_layout()
plt.show()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(
    max_features=5000,
    stop_words="english",
    min_df=5,
    max_df=0.7,
    ngram_range=(1, 1)
)

X_tfidf = tfidf.fit_transform(df["clean_text"])

print(X_tfidf.shape)
print("نسبة الأصفار (Sparsity):", 1 - X_tfidf.nnz / (X_tfidf.shape[0] * X_tfidf.shape[1]))
```

```python
from sklearn.decomposition import TruncatedSVD
import numpy as np

svd = TruncatedSVD(n_components=100, random_state=42)
X_svd = svd.fit_transform(X_tfidf)

print(X_svd.shape)
print("نسبة الـ Variance المحفوظة:", svd.explained_variance_ratio_.sum())

# رسم الـ Variance التراكمي عشان نتأكد إن 100 بعد كافي
cumulative_variance = np.cumsum(svd.explained_variance_ratio_)
plt.figure(figsize=(8, 4))
plt.plot(range(1, 101), cumulative_variance, marker=".")
plt.xlabel("عدد الأبعاد")
plt.ylabel("Variance التراكمي المحفوظ")
plt.title("Explained Variance — Truncated SVD")
plt.grid(True)
plt.show()
```

```python
for n in [50, 100, 200, 300, 500]:
    svd_test = TruncatedSVD(n_components=n, random_state=42)
    svd_test.fit(X_tfidf)
    print(f"n_components={n}: Variance = {svd_test.explained_variance_ratio_.sum():.4f}")
```

```python
svd = TruncatedSVD(n_components=300, random_state=42)
X_svd = svd.fit_transform(X_tfidf)

print(X_svd.shape)
print("نسبة الـ Variance المحفوظة:", svd.explained_variance_ratio_.sum())
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

inertias, silhouettes = [], []
K_range = range(5, 26, 2)
for k in K_range:
    km = KMeans(n_clusters=k, init="k-means++", n_init=10, random_state=42)
    labels = km.fit_predict(X_svd)
    inertias.append(km.inertia_)
    silhouettes.append(silhouette_score(X_svd, labels))
    print(f"K={k}: Silhouette={silhouettes[-1]:.4f}")

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].plot(list(K_range), inertias, marker="o")
axes[0].set_title("Elbow Method")
axes[1].plot(list(K_range), silhouettes, marker="o", color="orange")
axes[1].set_title("Silhouette Score")
plt.tight_layout()
plt.show()
```

```python
from sklearn.preprocessing import normalize

X_svd_norm = normalize(X_svd, norm="l2")

# نعيد نفس تجربة K لكن على النسخة المطبّعة
silhouettes_norm = []
K_range = range(5, 26, 2)
for k in K_range:
    km = KMeans(n_clusters=k, init="k-means++", n_init=10, random_state=42)
    labels = km.fit_predict(X_svd_norm)
    silhouettes_norm.append(silhouette_score(X_svd_norm, labels))
    print(f"K={k}: Silhouette={silhouettes_norm[-1]:.4f}")
```

```python
from sklearn.metrics import adjusted_rand_score, normalized_mutual_info_score

km20 = KMeans(n_clusters=20, init="k-means++", n_init=10, random_state=42)
labels_20 = km20.fit_predict(X_svd_norm)

ari = adjusted_rand_score(df["category"], labels_20)
nmi = normalized_mutual_info_score(df["category"], labels_20)
print("ARI:", ari)
print("NMI:", nmi)
```

```python
from sklearn.decomposition import LatentDirichletAllocation
from sklearn.feature_extraction.text import CountVectorizer

# LDA بيشتغل أحسن مع Raw Counts مش TF-IDF (مش هنستخدم X_tfidf هنا)
count_vec = CountVectorizer(max_features=5000, stop_words="english", min_df=5, max_df=0.7)
X_counts = count_vec.fit_transform(df["clean_text"])

lda = LatentDirichletAllocation(n_components=20, random_state=42, max_iter=10, learning_method="batch")
lda_topics = lda.fit_transform(X_counts)

# أهم 10 كلمات لكل Topic
feature_names = count_vec.get_feature_names_out()
for topic_idx, topic in enumerate(lda.components_):
    top_words = [feature_names[i] for i in topic.argsort()[-10:][::-1]]
    print(f"Topic {topic_idx}: {', '.join(top_words)}")
```

```python
def is_gibberish(text, threshold=0.12):
    """
    بيكتشف نصوص الـ uuencoded عن طريق نسبة الكلمات القصيرة جدًا
    (1-2 حرف) اللي مش موجودة في قاموس حقيقي — نمط مميز لترميز uuencode.
    """
    words = text.split()
    if len(words) < 5:
        return True
    short_words = [w for w in words if len(w) <= 2]
    return (len(short_words) / len(words)) > threshold

df["is_gibberish"] = df["clean_text"].apply(is_gibberish)
print(df["is_gibberish"].sum(), "مستند مشبوه")
print(df[df["is_gibberish"]]["category"].value_counts().head(10))

# نشيلهم ونعيد بناء كل حاجة من CountVectorizer
df_clean2 = df[~df["is_gibberish"]].reset_index(drop=True)
print("الحجم بعد الفلترة:", df_clean2.shape)
```

```python
from collections import Counter

def is_gibberish_v2(text, threshold=0.25):
    words = text.split()
    if len(words) < 10:
        return False  # نص قصير جدًا، مش كفاية نحكم عليه
    counts = Counter(words)
    most_common_word, most_common_count = counts.most_common(1)[0]
    # لو أكتر كلمة متكررة (وطولها قصير) بتاخد نسبة كبيرة جدًا من المستند
    return len(most_common_word) <= 3 and (most_common_count / len(words)) > threshold

df["is_gibberish"] = df["clean_text"].apply(is_gibberish_v2)
print(df["is_gibberish"].sum(), "مستند مشبوه")
print(df[df["is_gibberish"]]["category"].value_counts().head(10))

df_clean2 = df[~df["is_gibberish"]].reset_index(drop=True)
print("الحجم بعد الفلترة:", df_clean2.shape)
```

```python
# نشيل الـ13 مستند من الداتا الرئيسية
df = df[~df["is_gibberish"]].reset_index(drop=True)
print("الحجم النهائي:", df.shape)

# نعيد بناء CountVectorizer وLDA على الداتا النضيفة
count_vec = CountVectorizer(max_features=5000, stop_words="english", min_df=5, max_df=0.7)
X_counts = count_vec.fit_transform(df["clean_text"])

lda = LatentDirichletAllocation(n_components=20, random_state=42, max_iter=10, learning_method="batch")
lda_topics = lda.fit_transform(X_counts)

feature_names = count_vec.get_feature_names_out()
for topic_idx, topic in enumerate(lda.components_):
    top_words = [feature_names[i] for i in topic.argsort()[-10:][::-1]]
    print(f"Topic {topic_idx}: {', '.join(top_words)}")
```

```python
import numpy as np

# أهم Topic لكل مستند (اللي نسبته أعلى حاجة)
dominant_topic = lda_topics.argmax(axis=1)

ari_lda = adjusted_rand_score(df["category"], dominant_topic)
nmi_lda = normalized_mutual_info_score(df["category"], dominant_topic)
print("LDA — ARI:", ari_lda)
print("LDA — NMI:", nmi_lda)
```

```python
# إعادة بناء TF-IDF و SVD على الداتا بعد فلترة الـ uuencoded (16,752 صف)
tfidf = TfidfVectorizer(max_features=5000, stop_words="english", min_df=5, max_df=0.7, ngram_range=(1, 1))
X_tfidf = tfidf.fit_transform(df["clean_text"])

svd = TruncatedSVD(n_components=300, random_state=42)
X_svd = svd.fit_transform(X_tfidf)

X_svd_norm = normalize(X_svd, norm="l2")

print("X_svd_norm:", X_svd_norm.shape)
print("df:", df.shape)

# وبعدين نعيد تدريب K-Means عشان labels_20 يبقى بنفس الحجم كمان
km20 = KMeans(n_clusters=20, init="k-means++", n_init=10, random_state=42)
labels_20 = km20.fit_predict(X_svd_norm)
print("labels_20:", labels_20.shape)
```

```python
reducer_2d = umap.UMAP(n_neighbors=15, min_dist=0.1, n_components=2, random_state=42)
X_umap_2d = reducer_2d.fit_transform(X_svd_norm)

fig, axes = plt.subplots(1, 2, figsize=(18, 8))

scatter1 = axes[0].scatter(X_umap_2d[:, 0], X_umap_2d[:, 1], c=pd.factorize(df["category"])[0],
                            cmap="tab20", s=5, alpha=0.6)
axes[0].set_title("UMAP — الفئات الحقيقية (Ground Truth)")

scatter2 = axes[1].scatter(X_umap_2d[:, 0], X_umap_2d[:, 1], c=labels_20,
                            cmap="tab20", s=5, alpha=0.6)
axes[1].set_title("UMAP — تصنيف K-Means")

plt.tight_layout()
plt.show()
```

```python
def predict_text(new_text):
    # نفس خطوة التنضيف بالظبط اللي طبقناها وقت التدريب
    clean = clean_text(new_text)

    # ===== K-Means (عبر TF-IDF + SVD + Normalize) =====
    new_tfidf = tfidf.transform([clean])        # transform مش fit_transform!
    new_svd = svd.transform(new_tfidf)
    new_svd_norm = normalize(new_svd, norm="l2")
    cluster_id = km20.predict(new_svd_norm)[0]

    # ===== LDA (عبر CountVectorizer) =====
    new_counts = count_vec.transform([clean])
    topic_dist = lda.transform(new_counts)[0]
    topic_id = topic_dist.argmax()
    topic_confidence = topic_dist[topic_id]

    return {
        "cluster_kmeans": int(cluster_id),
        "topic_lda": int(topic_id),
        "topic_name": topic_names.get(topic_id, f"Topic {topic_id}"),
        "topic_confidence": round(float(topic_confidence), 3),
    }

# تجربة
sample = "NASA announced a new mission to launch a satellite into orbit around the moon."
print(predict_text(sample))
```
